```python

print("Họ tên: PhanCongTri")
print("Mã số sinh viên: 6451071080")

# Linear algebra
import numpy as np

# Data processing
import pandas as pd

# Data visualization
import seaborn as sns
%matplotlib inline
from matplotlib import pyplot as plt
from matplotlib import style

# Đọc dữ liệu từ file bank-full(1).csv
df = pd.read_csv(
    "bank-full.csv",
    sep=";",
    dtype={
        "age": np.int64,
        "balance": np.int64,
        "day": np.int64,
        "duration": np.int64,
        "campaign": np.int64,
        "pdays": np.int64,
        "previous": np.int64
    }
)

```

```python
print("Họ tên: PhanCongTri")
print("Mã số sinh viên: 6451071080")

# Chuyển đổi cột mục tiêu 'y' thành dạng số (1 cho 'yes', 0 cho 'no')
df['survived_num'] = df['y'].apply(lambda x: 1 if x == 'yes' else 0)

survived = 'yes (subscribed)'
not_survived = 'no (not subscribed)'

fig, axes = plt.subplots(nrows=1, ncols=2, figsize=(10, 4))
group1 = df[df['housing'] == 'yes']
group2 = df[df['housing'] == 'no']

# Biểu đồ cho nhóm có vay mua nhà (Housing: Yes)
ax = sns.histplot(group1[group1['survived_num'] == 1].age.dropna(), bins=18, label=survived, ax=axes[0], kde=False, color='g')
ax = sns.histplot(group1[group1['survived_num'] == 0].age.dropna(), bins=40, label=not_survived, ax=axes[0], kde=False, color='y')
ax.legend()
ax.set_title('Housing: Yes')

# Biểu đồ cho nhóm không vay mua nhà (Housing: No)
ax = sns.histplot(group2[group2['survived_num'] == 1].age.dropna(), bins=18, label=survived, ax=axes[1], kde=False, color='g')
ax = sns.histplot(group2[group2['survived_num'] == 0].age.dropna(), bins=40, label=not_survived, ax=axes[1], kde=False, color='y')
ax.legend()
_ = ax.set_title('Housing: No')

plt.tight_layout()
plt.show()
```
