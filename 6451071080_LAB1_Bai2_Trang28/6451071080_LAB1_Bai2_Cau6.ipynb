```python

print("Họ tên: PhanCongTri")
print("Mã số sinh viên: 6451071080")

# Linear algebra
import numpy as np

# Data processing
import pandas as pd

# Data visualization
import seaborn as sns
%matplotlib inline
from matplotlib import pyplot as plt
from matplotlib import style

# Đọc dữ liệu từ file bank-full(1).csv
df = pd.read_csv(
    "bank-full.csv",
    sep=";",
    dtype={
        "age": np.int64,
        "balance": np.int64,
        "day": np.int64,
        "duration": np.int64,
        "campaign": np.int64,
        "pdays": np.int64,
        "previous": np.int64
    }
)

```

```python
print("Họ tên: PhanCongTri")
print("Mã số sinh viên: 6451071080")

# Từ điển ánh xạ kết quả chiến dịch trước (poutcome) sang dạng số
deck = {"success": 1, "failure": 2, "other": 3, "unknown": 4}

for dataset in [df]:
    dataset['poutcome'] = dataset['poutcome'].fillna("unknown")
    
    # Ánh xạ cột poutcome sang dạng mã số poutcome_code
    dataset['poutcome_code'] = dataset['poutcome'].map(deck)
    dataset['poutcome_code'] = dataset['poutcome_code'].fillna(0)
    dataset['poutcome_code'] = dataset['poutcome_code'].astype(int)

# Kiểm tra phân phối giá trị sau khi chuyển đổi
print("Phân phối giá trị của poutcome_code:")
print(df['poutcome_code'].value_counts())
```

```python
print("Họ tên: PhanCongTri")
print("Mã số sinh viên: 6451071080")

df = df.drop(['contact'], axis=1)
df = df.drop(['day'], axis=1)
df = df.drop(['month'], axis=1)
df = df.drop(['poutcome'], axis=1)
```

```python
print("Họ tên: PhanCongTri")
print("Mã số sinh viên: 6451071080")

for dataset in [df]:
    mean = df["age"].mean()
    std = df["age"].std()
    is_null = dataset["age"].isnull().sum()
    # compute random numbers between the mean, std and is_null
    rand_age = np.random.randint(mean - std, mean + std, size=is_null)
    # fill NaN values in Age column with random values generated
    age_slice = dataset["age"].copy()
    age_slice[np.isnan(age_slice)] = rand_age
    dataset["age"] = age_slice
    dataset["age"] = dataset["age"].astype(int)

df["age"].isnull().sum()
```
