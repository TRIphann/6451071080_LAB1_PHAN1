```python

print("Họ tên: PhanCongTri")
print("Mã số sinh viên: 6451071080")

# Linear algebra
import numpy as np

# Data processing
import pandas as pd

# Data visualization
import seaborn as sns
%matplotlib inline
from matplotlib import pyplot as plt
from matplotlib import style

# Đọc dữ liệu từ file bank-full(1).csv
df = pd.read_csv(
    "bank-full.csv",
    sep=";",
    dtype={
        "age": np.int64,
        "balance": np.int64,
        "day": np.int64,
        "duration": np.int64,
        "campaign": np.int64,
        "pdays": np.int64,
        "previous": np.int64
    }
)

```

```python
print("Họ tên: PhanCongTri")
print("Mã số sinh viên: 6451071080")

# Chuyển đổi cột mục tiêu 'y' thành dạng số (1 cho 'yes', 0 cho 'no')
df['survived_num'] = df['y'].apply(lambda x: 1 if x == 'yes' else 0)

# Xây dựng FacetGrid tương đương trên tập dữ liệu ngân hàng
FacetGrid = sns.FacetGrid(df, row='contact', height=4.5, aspect=1.6)
FacetGrid.map(sns.pointplot, 'education', 'survived_num', 'housing', palette=None, order=None, hue_order=None, errorbar=None)
FacetGrid.add_legend()

plt.tight_layout()
plt.show()
```

```python
print("Họ tên: PhanCongTri")
print("Mã số sinh viên: 6451071080")

# Chuyển đổi cột mục tiêu 'y' thành dạng số (1 cho 'yes', 0 cho 'no')
df['survived_num'] = df['y'].apply(lambda x: 1 if x == 'yes' else 0)

# Lọc bỏ giá trị 'unknown' ở cột education để chỉ giữ lại 3 mức học vấn: primary, secondary, tertiary
df_filtered = df[df['education'] != 'unknown']

# Vẽ biểu đồ cột với đúng 3 cột tương ứng với 3 mức học vấn và 3 màu sắc khác nhau
plt.figure(figsize=(8, 6))
sns.barplot(
    x='education', 
    y='survived_num', 
    hue='education', 
    data=df_filtered, 
    palette=['blue', 'orange', 'green'], 
    legend=False
)

plt.title('Bank Subscription Rate by Education Level (Filtered 3 Levels)')
plt.xlabel('Education')
plt.ylabel('Subscription Rate')
plt.tight_layout()
plt.show()
```
