```python
print("Họ tên: PhanCongTri")
print("Mã số sinh viên: 6451071080")

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Đọc dữ liệu từ file đã gửi
df = pd.read_csv('tb_tracking_processed.csv')

# Định nghĩa các nhãn tương đương
group_1_label = 'high_temp'      
group_2_label = 'low_temp'       

fig, axes = plt.subplots(nrows=1, ncols=2, figsize=(10, 4))

group1 = df[df['device_id'] == 99005]
group2 = df[df['device_id'] != 99005]

ax = sns.histplot(group1[group1['temperature'] > 2000000].temperature.dropna(), bins=18, label=group_1_label, ax=axes[0], kde=False, color='g')
ax = sns.histplot(group1[group1['temperature'] <= 2000000].temperature.dropna(), bins=40, label=group_2_label, ax=axes[0], kde=False, color='y')
ax.legend()
ax.set_title('Device 99005')

ax = sns.histplot(group2[group2['temperature'] > 2000000].temperature.dropna(), bins=18, label=group_1_label, ax=axes[1], kde=False, color='g')
ax = sns.histplot(group2[group2['temperature'] <= 2000000].temperature.dropna(), bins=40, label=group_2_label, ax=axes[1], kde=False, color='y')
ax.legend()
_ = ax.set_title('Other Devices')

plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(nrows=3, ncols=1, figsize=(11, 10))
sns.scatterplot(x=df['minutes'], y=df['co_level'], ax=axes[0], color='blue', s=12, alpha=0.6)
axes[0].set_title('CO Level vs Minutes')
axes[0].set_xlabel('Minutes')
axes[0].set_ylabel('CO Level')
sns.scatterplot(x=df['minutes'], y=df['humidity'], ax=axes[1], color='orange', s=12, alpha=0.6)
axes[1].set_title('Humidity vs Minutes')
axes[1].set_xlabel('Minutes')
axes[1].set_ylabel('Humidity')
sns.scatterplot(x=df['minutes'], y=df['temperature'], ax=axes[2], color='green', s=12, alpha=0.6)
axes[2].set_title('Temperature vs Minutes')
axes[2].set_xlabel('Minutes')
axes[2].set_ylabel('Temperature')
plt.tight_layout()
plt.show()
```
