```python

print("Họ tên: PhanCongTri")
print("Mã số sinh viên: 6451071080")

# Linear algebra
import numpy as np

# Data processing
import pandas as pd

# Data visualization
import seaborn as sns
%matplotlib inline
from matplotlib import pyplot as plt
from matplotlib import style

# ============================================================
# a. Đọc dữ liệu và đặt tên cột
# ============================================================
# Kiểm tra: file tb_tracking.csv KHÔNG có header (dòng đầu là dữ liệu số)
# 6 cột theo thứ tự: id, device_id, co_level, humidity, temperature, time

col_names = ['id', 'device_id', 'co_level', 'humidity', 'temperature', 'time']

df = pd.read_csv(
    'tb_tracking.csv',
    header=None,          # file không có header
    names=col_names,      # gán tên cột theo thứ tự yêu cầu
    encoding='utf-8'
)

print(f"Số cột: {df.shape[1]}")
print(f"Tên cột: {list(df.columns)}")
print(f"Tổng số dòng ban đầu: {len(df)}")
df.head()
```

```python
print("Họ tên: PhanCongTri")
print("Mã số sinh viên: 6451071080")

# Xem thông tin kiểu dữ liệu ban đầu
df.info()
```

```python
print("Họ tên: PhanCongTri")
print("Mã số sinh viên: 6451071080")

# ============================================================
# b. Lọc thời gian: giữ dữ liệu từ ngày 03/01/2018 trở đi
# (tức ngày 3 tháng 1 năm 2018)
# ============================================================

# Chuyển cột time sang kiểu datetime
df['time'] = pd.to_datetime(df['time'], errors='coerce', format='mixed')

so_dong_truoc = len(df)
print(f"Số dòng TRƯỚC khi lọc: {so_dong_truoc}")

# Loại bỏ dữ liệu trước ngày 03/01/2018
ngay_gioi_han = pd.Timestamp('2018-01-03')
df = df[df['time'] >= ngay_gioi_han]

so_dong_sau = len(df)
print(f"Số dòng SAU khi lọc (từ 03/01/2018 trở đi): {so_dong_sau}")
print(f"Số dòng đã loại bỏ: {so_dong_truoc - so_dong_sau}")
df.head()
```

```python
print("Họ tên: PhanCongTri")
print("Mã số sinh viên: 6451071080")

# ============================================================
# c. Tạo cột minutes = giờ * 60 + phút
# Ví dụ: 08:30 -> 510, 12:45 -> 765
# ============================================================

df['minutes'] = df['time'].dt.hour * 60 + df['time'].dt.minute

print("Cột minutes đã được tạo.")
print(f"Ví dụ: giờ 8:30 -> {8*60+30} phút | 12:45 -> {12*60+45} phút")
df[['time', 'minutes']].head(10)
```

```python
print("Họ tên: PhanCongTri")
print("Mã số sinh viên: 6451071080")

# ============================================================
# Lưu dataset đã xử lý thành file mới (không ghi đè gốc)
# ============================================================

output_file = 'tb_tracking_processed.csv'
df.to_csv(output_file, index=False, encoding='utf-8')
print(f"Dataset đã xử lý được lưu vào: {output_file}")
print(f"Số dòng: {len(df)}, Số cột: {df.shape[1]}")
print(f"Cột: {list(df.columns)}")
df.describe()
```
