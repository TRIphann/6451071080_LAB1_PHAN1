```python

print("Họ tên: PhanCongTri")
print("Mã số sinh viên: 6451071080")

# Linear algebra
import numpy as np

# Data processing
import pandas as pd

# Data visualization
import seaborn as sns
%matplotlib inline
from matplotlib import pyplot as plt
from matplotlib import style

# Đọc dữ liệu đã xử lý từ file tb_tracking_processed.csv
df = pd.read_csv(
    'tb_tracking_processed.csv',
    parse_dates=['time']
)
```

```python
print("Họ tên: PhanCongTri")
print("Mã số sinh viên: 6451071080")

import re

# Tạo từ điển ánh xạ thiết bị tương đương với deck
deck = {"DeviceA": 1, "DeviceB": 2, "DeviceC": 3}

for dataset in [df]:
    # Chuyển đổi device_id sang chuỗi định danh để áp dụng logic xử lý text giống code mẫu
    # device_id trong CSV dang float (1.0, 99005.0, 0.0) -> chuyen sang int truoc
    dataset['device_id_int'] = dataset['device_id'].fillna(0).astype(int)
    dataset['device_str'] = dataset['device_id_int'].astype(str)
    dataset['device_str'] = dataset['device_str'].replace({'1': 'DeviceA', '99005': 'DeviceB', '0': 'DeviceC'})
    
    # Điền giá trị thiếu và trích xuất chữ cái bằng biểu thức chính quy (giống hệt code mẫu)
    dataset['device_str'] = dataset['device_str'].fillna("DeviceU")
    dataset['deck'] = dataset['device_str'].map(lambda x: re.compile("([a-zA-Z]+)").search(x).group())
    dataset['deck'] = dataset['deck'].map(deck)
    dataset['deck'] = dataset['deck'].fillna(0)
    dataset['deck'] = dataset['deck'].astype(int)

# Xóa cột trung gian không cần thiết (tương đương drop cabin)
df = df.drop(['device_str'], axis=1)
```

```python
print("Họ tên: PhanCongTri")
print("Mã số sinh viên: 6451071080")

# Xóa các cột không cần thiết trong file tb_tracking (tương tự thao tác drop cột trong code mẫu)
df = df.drop(['id'], axis=1)
df = df.drop(['minutes'], axis=1)
```

```python
print("Họ tên: PhanCongTri")
print("Mã số sinh viên: 6451071080")

import numpy as np

for dataset in [df]:
    mean = dataset["temperature"].mean()
    std = dataset["temperature"].std()
    is_null = dataset["temperature"].isnull().sum()
    
    # Tính các giá trị ngẫu nhiên dựa trên mean, std và số lượng thiếu
    rand_val = np.random.randint(int(mean - std), int(mean + std) + 1, size = is_null)
    
    # Điền giá trị ngẫu nhiên vào các ô trống NaN của cột temperature
    temp_slice = dataset["temperature"].copy()
    temp_slice[temp_slice.isnull()] = rand_val
    dataset["temperature"] = temp_slice
    dataset["temperature"] = dataset["temperature"].astype(int)

df["temperature"].isnull().sum()
```
