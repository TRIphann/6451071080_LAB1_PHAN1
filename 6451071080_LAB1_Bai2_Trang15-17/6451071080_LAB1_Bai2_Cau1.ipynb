```python
print("Họ tên: PhanCongTri")
print("Mã số sinh viên: 6451071080")

# 5 dòng đầu tiên
display(df.head(12))

# Kiểu dữ liệu, số dòng không rỗng và bộ nhớ sử dụng
df.info()

# Thống kê mô tả các thuộc tính số: count, mean, std, min, các tứ phân vị, max
display(df.describe())

# Thống kê cả thuộc tính dạng chữ
display(df.describe(include="all").T)

# Bảng tổng hợp cơ bản cho từng thuộc tính
summary = pd.DataFrame({
    "Kieu_du_lieu": df.dtypes.astype(str),
    "So_gia_tri_khong_rong": df.notna().sum(),
    "So_gia_tri_rong": df.isna().sum(),
    "So_gia_tri_khac_nhau": df.nunique(dropna=True),
})
display(summary)
```
